```python

#Задача 4. Улитка

h, a, b = map(int, input().split())


effective_height = h - a


full_days = (effective_height + (a - b) - 1) // (a - b)


total_days = full_days + 1
print(total_days)

```

```python
#Задача 5. Электронные часы
def seconds_to_time(n):

    hours = n // 3600
    minutes = (n % 3600) // 60
    seconds = n % 60


    time_string = f"{hours}:{minutes:02}:{seconds:02}"
    return time_string

n = int(input("Введите количество секунд: "))

print(seconds_to_time(n))
```

```python
#Задача 6. Разность времен


def time_to_seconds(hours, minutes, seconds):
    return hours * 3600 + minutes * 60 + seconds

h1, m1, s1 = map(int, input("Введите первый момент времени (часы минуты секунды): ").split())

h2, m2, s2 = map(int, input("Введите второй момент времени (часы минуты секунды): ").split())


first_moment_seconds = time_to_seconds(h1, m1, s1)
second_moment_seconds = time_to_seconds(h2, m2, s2)


difference = second_moment_seconds - first_moment_seconds

print("разница в секундах - ", abs(difference))
```
